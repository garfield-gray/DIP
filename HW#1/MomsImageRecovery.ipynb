```python
#upload the file in advance
im = plt.imread("AnalogScan.jpg", ".jpg")
```

```python
print(im.shape)
```

```python
plt.figure(figsize = (6,10))
plt.imshow(im)
plt.show()
```

```python
#3/13/2023 let's start again
neg = 255*np.ones(im.shape, dtype=int) -im
plt.figure(figsize = (6,10))
plt.imshow(neg)
plt.show()
```

```python
#3/14/2023 an instance of matmul (tensor multiplication)
grayeq = neg@np.array([1/3,1/3,1/3])
graybetter = neg@np.array([0.2989,0.5870,0.114])
```

```python
f, axarr = plt.subplots(1,2,figsize=(17,15))
axarr[0].imshow(graybetter, cmap = 'gray')
axarr[1].imshow(grayeq, cmap = 'gray')
plt.show()
```

```python
grayeq[359,40],graybetter[359,40]
#as you can see, they're for sure different, but not that much!!
```

```python
#3/15/2023
f, axarr = plt.subplots(1,2,figsize=(17,5))
axarr[0].hist(grayeq.flatten(), bins = 300)
axarr[1].hist(graybetter.flatten(), bins = 300)
plt.show()
```

they're not that much different but the better one looks more equalized


```python
#making the equalized output
a = np.rint(graybetter.flatten()).astype(int)
b = np.rint(grayeq.flatten()).astype(int)

#accumilated
cumbetter = np.cumsum(np.bincount(a)/np.sum(np.bincount(a)))
cumeq = np.cumsum(np.bincount(b)/np.sum(np.bincount(b)))

#rounding the reults
#these Values maps the current pixels to the desired equal ones (look at the for)
outputValueBetter = np.rint(cumbetter*255).astype(int)
outputValueEq = np.rint(cumeq*255).astype(int)

graybetterEq = np.zeros(graybetter.shape, dtype=int)
grayeqEq = np.zeros(grayeq.shape, dtype=int)

#mapping them
for i in range(graybetter.shape[0]):
  for j in range(graybetter.shape[1]):
    graybetterEq[i,j] = outputValueBetter[np.rint(graybetter[i,j]).astype(int)]

for i in range(grayeq.shape[0]):
  for j in range(grayeq.shape[1]):
    grayeqEq[i,j] = outputValueEq[np.rint(grayeq[i,j]).astype(int)]
```

```python
f, axarr = plt.subplots(2,2,figsize=(17,15))
axarr[0,0].imshow(graybetter, cmap = 'gray')
axarr[0,1].imshow(grayeq, cmap = 'gray')
axarr[1,0].imshow(graybetterEq, cmap = 'gray')
axarr[1,1].imshow(grayeqEq, cmap = 'gray')
plt.show()
```

$$Better\;is\;better\;:)$$

```python
#3/15/2023
f, axarr = plt.subplots(2,2,figsize=(15,8))
axarr[0,0].hist(graybetter.flatten(), bins = 256)
axarr[0,1].hist(grayeq.flatten(), bins = 256)
axarr[1,0].hist(graybetterEq.flatten(), bins = 256)
axarr[1,1].hist(grayeqEq.flatten(), bins = 256)

plt.show()
```

I think that's exactly wha we want

```python
#upload the file in advance
im2 = plt.imread("1008_IM-0009-2001.dcm.png", ".png")
```

```python
print(im2.shape)
```

```python
print(im2[:,:,0])
print(im2[:,:,0]-im2[:,:,1])
print(im2[:,:,1]-im2[:,:,2])
print(im2[:,:,3])
```

```python
#3/21/2023
imm = im2[:,:,0]
plt.figure(figsize = (8,12))
plt.imshow(imm, cmap = "gray")
plt.show()
```

```python
#3/22/2023
#Average
m, n = imm.shape

win = np.ones([3,3], dtype=int)/9

immAveraged = np.zeros([m, n])

#it may take some time
for i in range(1, m-1):
  for j in range(1, n-1):
    newPix = imm[i-1, j-1]*win[0, 0]+imm[i-1, j]*win[0, 1]+imm[i-1, j + 1]*win[0, 2]+imm[i, j-1]*win[1, 0]+ imm[i, j]*win[1, 1]+imm[i, j + 1]*win[1, 2]+imm[i + 1, j-1]*win[2, 0]+imm[i + 1, j]*win[2, 1]+imm[i + 1, j + 1]*win[2, 2]
    immAveraged[i, j] = newPix
```

```python
f, axarr = plt.subplots(1,2,figsize=(17,15))
axarr[0].imshow(imm, cmap = 'gray')
axarr[1].imshow(immAveraged, cmap = 'gray')
plt.show()
```

```python
#Median
m, n = imm.shape

immMedian = np.zeros([m, n])

#it may take some time
for i in range(1, m-1):
  for j in range(1, n-1):
    immMedian[i, j] = np.median([imm[i-1, j-1],imm[i-1, j],imm[i-1, j + 1],imm[i, j-1],imm[i, j],imm[i, j + 1],imm[i + 1, j-1],imm[i + 1, j],imm[i + 1, j + 1]])
```

```python
f, axarr = plt.subplots(1,2,figsize=(17,15))
axarr[0].imshow(imm, cmap = 'gray')
axarr[1].imshow(immMedian, cmap = 'gray')
plt.show()
#median works very well!!
```

```python
f, axarr = plt.subplots(1,2,figsize=(17,5))
axarr[0].hist(imm.flatten(), bins = 300)
axarr[1].hist(immMedian.flatten(), bins = 300)
plt.show()
```

```python
#Max
m, n = imm.shape

immMax = np.zeros([m, n])

#it may take some time
for i in range(1, m-1):
  for j in range(1, n-1):
    immMax[i, j] = np.max([imm[i-1, j-1],imm[i-1, j],imm[i-1, j + 1],imm[i, j-1],imm[i, j],imm[i, j + 1],imm[i + 1, j-1],imm[i + 1, j],imm[i + 1, j + 1]])
```

```python
f, axarr = plt.subplots(1,2,figsize=(17,15))
axarr[0].imshow(imm, cmap = 'gray')
axarr[1].imshow(immMax, cmap = 'gray')
plt.show()
#Max is aweful!!
```

```python
#Min
m, n = imm.shape

immMin = np.zeros([m, n])

#it may take some time
for i in range(1, m-1):
  for j in range(1, n-1):
    immMin[i, j] = np.min([imm[i-1, j-1],imm[i-1, j],imm[i-1, j + 1],imm[i, j-1],imm[i, j],imm[i, j + 1],imm[i + 1, j-1],imm[i + 1, j],imm[i + 1, j + 1]])
```

```python
f, axarr = plt.subplots(1,2,figsize=(17,15))
axarr[0].imshow(imm, cmap = 'gray')
axarr[1].imshow(immMin, cmap = 'gray')
plt.show()
#Min is Hell of a Deal!!
#Even better than Median
```

```python
f, axarr = plt.subplots(1,2,figsize=(17,5))
axarr[0].hist(imm.flatten(), bins = 300)
axarr[1].hist(immMin.flatten(), bins = 300)
plt.show()
```

```python
#Gamma Correction
gamma = 0.5
immGamma = imm  ** gamma

```

```python
f, axarr = plt.subplots(1,2,figsize=(17,15))
axarr[0].imshow(imm, cmap = 'gray')
axarr[1].imshow(immGamma, cmap = 'gray')
plt.show()
#Gamma Doesn't help npise Reduction (all values)
```

```python
#Min and Gamma

immGammaMin = immMin **1.1
#tried many values for gamma and the best one happend in this one
```

```python
f, axarr = plt.subplots(1,2,figsize=(17,15))
axarr[0].imshow(immMin, cmap = 'gray')
axarr[1].imshow(immGammaMin, cmap = 'gray')
plt.show()
#I beleive these two work very well!
```

```python
f, axarr = plt.subplots(1,2,figsize=(17,5))
axarr[0].hist(immMin.flatten(), bins = 300)
axarr[1].hist(immGammaMin.flatten(), bins = 300)
plt.show()
```

$histograms\; are\;still\;not\;the\;way\;we\;want$
\
$then\;we'll\;equalize\;them$

```python
#making the equalized output
aa = np.rint((immGammaMin*255).flatten()).astype(int)

#accumilated
cum = np.cumsum(np.bincount(aa)/np.sum(np.bincount(aa)))

#rounding the reults
#these Values maps the current pixels to the desired equal ones (look at the for)
output = np.rint(cum*255).astype(int)

immEq = np.zeros(immGammaMin.shape, dtype=int)

#mapping them
for i in range(immGammaMin.shape[0]):
  for j in range(immGammaMin.shape[1]):
    immEq[i,j] = output[np.rint((immGammaMin[i,j]*255)).astype(int)]
```

```python
f, axarr = plt.subplots(1,2,figsize=(17,15))
axarr[0].imshow(immMin, cmap = 'gray')
axarr[1].imshow(immEq, cmap = 'gray')
plt.show()
#I beleive these two work very well!
```

```python
#min filter one more time
m, n = immEq.shape

immMinMin = np.zeros([m, n])

#it may take some time
for i in range(1, m-1):
  for j in range(1, n-1):
    immMinMin[i, j] = np.min([immEq[i-1, j-1],immEq[i-1, j],immEq[i-1, j + 1],immEq[i, j-1],immEq[i, j],immEq[i, j + 1],immEq[i + 1, j-1],immEq[i + 1, j],immEq[i + 1, j + 1]])
```

```python
f, axarr = plt.subplots(1,2,figsize=(17,15))
axarr[0].imshow(immEq, cmap = 'gray')
axarr[1].imshow(immMinMin, cmap = 'gray')
plt.show()
```
